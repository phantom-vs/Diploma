```python
def has_nonempty(intervals):
    return bool(intervals) and any(float(x["end"]) > float(x["start"]) for x in intervals)

rows = []
for idx in segment_loader.iter_segment_indices(export_root):
    meta = segment_loader.load_segment_meta(export_root, idx)
    rows.append({
        "seg": idx,
        "sleep": has_nonempty(meta.get("intervals_sleep", [])),
        "wake": has_nonempty(meta.get("intervals_wake", [])),
        "depd_by_ch": has_nonempty(meta.get("intervals_by_channel", [])),
        "depd_global": has_nonempty(meta.get("intervals_global", [])),
        "n_sleep_intervals": len(meta.get("intervals_sleep", [])),
        "n_wake_intervals": len(meta.get("intervals_wake", [])),
        "n_depd_by_ch": len(meta.get("intervals_by_channel", [])),
        "n_depd_global": len(meta.get("intervals_global", [])),
    })

rows[:10]
```
